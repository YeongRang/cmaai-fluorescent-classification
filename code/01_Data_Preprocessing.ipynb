```python
# Canonical SMILES Function
def to_canonical(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol:
        return Chem.MolToSmiles(mol, canonical=True)
    else:
        return None
```

```python
# Non-Fluorescent
df_N = pd.read_csv('nonfluorescent.csv', encoding='utf-8-sig')

df_N['SMILES'] = df_N['SMILES'].apply(to_canonical)
df_N = df_N.drop_duplicates(subset='SMILES', keep='first')
df_N = df_N.reset_index(drop=True)

display(df_N)
```

```python
# Fluorescent
df_F = pd.read_csv('fluorescent.csv', encoding='utf-8-sig')

df_F['SMILES'] = df_F['SMILES'].apply(to_canonical)
df_F = df_F.drop_duplicates(subset='SMILES', keep='first')
df_F = df_F.reset_index(drop=True)

display(df_F)
```

```python
# Non-Fluorescent Class Labelling
df_ndye_selected = df_N[['SMILES']].copy()
df_ndye_selected['Class'] = 0
# df_ndye_selected['Class'] = 'Non-Fluorescent'
# display(df_ndye_selected)

# Fluorescent Class Labelling
df_fdye_selected = df_F[['SMILES']].copy()
df_fdye_selected['Class'] = 1
# df_fdye_selected['Class'] = 'Fluorescent'
# display(df_fdye_selected)

# Total Dataset
df = pd.concat([df_ndye_selected, df_fdye_selected], ignore_index=True)

# 결과 확인
print(f"Total Data: {len(df)}")
print("Shape:", df.shape)
display(df)
```

```python
df.to_csv('fluorclass_dataset.csv',index=False,encoding='utf-8-sig')
```
